# GameForge AI: Action & Pose Conditional Diffusion Training
### Module 4: 10,000 Paired Samples — Dual GPU Accelerated (Kaggle T4 x2)
**With Explicit Action-ID Conditioning (Walk, Slash, Thrust, Spellcast, Hurt)**

---
### Why this Architecture Truly Articulates Different Poses:
1. **Reference Conditioning ($x_{\text{ref}}$)**: Injects character identity, skin, clothing, and armor palette.
2. **Action-ID Embedding ($a \in \{0..15\}$)**: Tells the UNet exactly **which combat action & direction** to generate.
3. **Dual Condition Fusion**: Combines sinusoidal timestep embeddings $t$ with learned action embedding vectors $\mathbf{e}_a$, preventing mode collapse to the static average.
4. **Kaggle Dataset Auto-Detection**: Checks local Kaggle input directory `/kaggle/input/gameforge-lpc-animation-deformation` first, and falls back to HuggingFace streaming seamlessly!

```python
# Cell 1: Environment Setup & Dual GPU Detection
import os
import sys
import time
import json
from pathlib import Path
import numpy as np
import pandas as pd
from PIL import Image
import matplotlib.pyplot as plt
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
NUM_GPUS = torch.cuda.device_count()
print(f"⚡ PyTorch Version: {torch.__version__}")
print(f"⚡ Compute Target: {DEVICE} ({NUM_GPUS} GPUs Available)")
if NUM_GPUS > 0:
    for i in range(NUM_GPUS):
        print(f"   GPU {i}: {torch.cuda.get_device_name(i)}")

WORK = Path("/kaggle/working") if Path("/kaggle/working").exists() else Path("./diffusion_workspace")
WORK.mkdir(parents=True, exist_ok=True)
CHECKPOINT_DIR = WORK / "checkpoints"
CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)
print("Working directory:", WORK)
```

```python
# Cell 2: Dataset Loading (Kaggle Add-Dataset OR HuggingFace Fallback)
KAGGLE_INPUT = Path("/kaggle/input")
dataset_dirs = list(KAGGLE_INPUT.glob("*lpc*")) + list(KAGGLE_INPUT.glob("*gameforge*"))

USE_LOCAL_KAGGLE = len(dataset_dirs) > 0
print(f"Detected Kaggle Inputs: {dataset_dirs}")

if not USE_LOCAL_KAGGLE:
    print("Loading via HuggingFace streaming...")
    !pip install -q datasets huggingface_hub diffusers torchvision
    from datasets import load_dataset
    action_ds = load_dataset("carlosuperb/lpc-action-pixel-art-diffusion", split="train")
    fourview_ds = load_dataset("carlosuperb/lpc-4view-pixel-art-diffusion", split="train")
    print(f"HuggingFace Datasets Ready: Action={len(action_ds):,} | 4-View={len(fourview_ds):,}")
else:
    print(f"Using attached Kaggle Dataset at: {dataset_dirs[0]}")
```

```python
# Cell 3: Action-Conditioned Paired Dataset Construction (10,000 Pairs)
GRID_COLS = 9
GRID_ROWS = 16
IMAGE_SIZE = 64
MAX_CHARACTERS = 625 # 625 chars * 16 action poses = 10,000 paired samples

# 16 Explicit Action Categories mapped to Grid Rows/Cols:
# Row 0-3: Spellcast (Up, Left, Down, Right)
# Row 4-7: Thrust (Up, Left, Down, Right)
# Row 8-11: Walk (Up, Left, Down, Right)
# Row 12-15: Slash (Up, Left, Down, Right)
ACTION_NAMES = [
    "Spellcast_Up", "Spellcast_Left", "Spellcast_Down", "Spellcast_Right",
    "Thrust_Up", "Thrust_Left", "Thrust_Down", "Thrust_Right",
    "Walk_Up", "Walk_Left", "Walk_Down", "Walk_Right",
    "Slash_Up", "Slash_Left", "Slash_Down", "Slash_Right"
]

pairs = []
print(f"Extracting 10,000 action-conditioned tuples from {MAX_CHARACTERS} characters...")

if not USE_LOCAL_KAGGLE:
    limit = min(MAX_CHARACTERS, len(action_ds), len(fourview_ds))
    for idx in tqdm(range(limit), desc="Processing LPC Sheets"):
        try:
            ref_im = fourview_ds[idx]["image"].convert("RGBA").resize((IMAGE_SIZE, IMAGE_SIZE), Image.Resampling.NEAREST)
            act_sheet = action_ds[idx]["image"].convert("RGBA")
            cw = act_sheet.width // GRID_COLS
            ch = act_sheet.height // GRID_ROWS
            
            ref_t = torch.from_numpy(np.asarray(ref_im, dtype=np.float32) / 127.5 - 1.0).permute(2, 0, 1)[:3]
            
            # Extract 16 distinct action poses per character
            action_id = 0
            for r in range(min(16, GRID_ROWS)):
                c = 2 # Active stride/strike column
                box = (c * cw, r * ch, (c + 1) * cw, (r + 1) * ch)
                crop = act_sheet.crop(box)
                if np.asarray(crop.getchannel("A")).sum() < 40:
                    continue
                crop_64 = crop.resize((IMAGE_SIZE, IMAGE_SIZE), Image.Resampling.NEAREST)
                tgt_t = torch.from_numpy(np.asarray(crop_64, dtype=np.float32) / 127.5 - 1.0).permute(2, 0, 1)[:3]
                pairs.append((ref_t, tgt_t, action_id % 16))
                action_id += 1
        except Exception:
            continue
else:
    # Fast extraction from Kaggle attached local images
    data_root = dataset_dirs[0]
    all_pngs = sorted(list(data_root.glob("**/*.png")))
    print(f"Found {len(all_pngs)} images in local dataset!")
    limit = min(MAX_CHARACTERS, len(all_pngs))
    for idx in tqdm(range(limit), desc="Slicing Local Sheets"):
        try:
            sheet = Image.open(all_pngs[idx]).convert("RGBA")
            cw = sheet.width // GRID_COLS if sheet.width >= 576 else sheet.width // 4
            ch = sheet.height // GRID_ROWS if sheet.height >= 768 else sheet.height // 4
            ref_crop = sheet.crop((0, 0, cw, ch)).resize((IMAGE_SIZE, IMAGE_SIZE), Image.Resampling.NEAREST)
            ref_t = torch.from_numpy(np.asarray(ref_crop, dtype=np.float32) / 127.5 - 1.0).permute(2, 0, 1)[:3]
            
            for a_idx in range(16):
                r = a_idx % (sheet.height // ch)
                c = min(2, (sheet.width // cw) - 1)
                box = (c * cw, r * ch, (c + 1) * cw, (r + 1) * ch)
                crop = sheet.crop(box).resize((IMAGE_SIZE, IMAGE_SIZE), Image.Resampling.NEAREST)
                tgt_t = torch.from_numpy(np.asarray(crop, dtype=np.float32) / 127.5 - 1.0).permute(2, 0, 1)[:3]
                pairs.append((ref_t, tgt_t, a_idx))
        except Exception:
            continue

print(f"✅ Successfully paired {len(pairs):,} conditioned action training pairs!")

class ActionConditionedDataset(Dataset):
    def __init__(self, sample_list):
        self.samples = sample_list

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        ref, tgt, act_id = self.samples[idx]
        return {"reference": ref, "target": tgt, "action_id": torch.tensor(act_id, dtype=torch.long)}

dataset = ActionConditionedDataset(pairs)
PER_GPU_BATCH = 64
BATCH_SIZE = PER_GPU_BATCH * max(1, NUM_GPUS)
train_loader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True, pin_memory=True, drop_last=True)
print(f"DataLoader: Batch Size = {BATCH_SIZE} | Total Batches per Epoch = {len(train_loader)}")
```

```python
# Cell 4: True Action-Conditioned Diffusion U-Net
class SinusoidalPositionEmbeddings(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.dim = dim

    def forward(self, time):
        device = time.device
        half_dim = self.dim // 2
        embeddings = 10000 ** (torch.arange(half_dim, device=device).float() / (half_dim - 1))
        embeddings = time[:, None] * embeddings[None, :]
        embeddings = torch.cat((embeddings.sin(), embeddings.cos()), dim=-1)
        return embeddings

class ActionConditionedDiffusionUNet(nn.Module):
    def __init__(self, in_channels=6, out_channels=3, time_dim=128, num_actions=16, base=64):
        super().__init__()
        self.time_mlp = nn.Sequential(
            SinusoidalPositionEmbeddings(time_dim),
            nn.Linear(time_dim, time_dim),
            nn.SiLU()
        )
        # Action Condition Embedding (Tells model which frame/stance to generate)
        self.action_embed = nn.Embedding(num_actions, time_dim)
        self.cond_combine = nn.Linear(time_dim * 2, time_dim)
        
        self.in_conv = nn.Conv2d(in_channels, base, 3, padding=1)
        self.down1 = nn.Conv2d(base, base * 2, 4, 2, 1)
        self.time_emb1 = nn.Linear(time_dim, base * 2)
        
        self.down2 = nn.Conv2d(base * 2, base * 4, 4, 2, 1)
        self.time_emb2 = nn.Linear(time_dim, base * 4)
        
        self.mid_conv = nn.Conv2d(base * 4, base * 4, 3, padding=1)
        
        self.up1 = nn.ConvTranspose2d(base * 4, base * 2, 4, 2, 1)
        self.up2 = nn.ConvTranspose2d(base * 2, base, 4, 2, 1)
        self.out_conv = nn.Conv2d(base, out_channels, 3, padding=1)
        
    def forward(self, noisy_target, timestep, reference, action_id):
        # Timestep + Action Embedding fusion
        t_emb = self.time_mlp(timestep)
        a_emb = self.action_embed(action_id)
        cond_emb = F.silu(self.cond_combine(torch.cat([t_emb, a_emb], dim=-1)))
        
        x = torch.cat([noisy_target, reference], dim=1) # (B, 6, 64, 64)
        
        h0 = F.silu(self.in_conv(x))
        h1 = F.silu(self.down1(h0) + self.time_emb1(cond_emb)[:, :, None, None])
        h2 = F.silu(self.down2(h1) + self.time_emb2(cond_emb)[:, :, None, None])
        
        m = F.silu(self.mid_conv(h2))
        
        u1 = F.silu(self.up1(m) + h1)
        u2 = F.silu(self.up2(u1) + h0)
        return self.out_conv(u2)

TIMESTEPS = 1000
betas = torch.linspace(1e-4, 0.02, TIMESTEPS).to(DEVICE)
alphas = 1.0 - betas
alphas_cumprod = torch.cumprod(alphas, dim=0).to(DEVICE)

raw_model = ActionConditionedDiffusionUNet(time_dim=128, base=64, num_actions=16).to(DEVICE)
total_p = sum(p.numel() for p in raw_model.parameters())
print(f"Action-Conditioned UNet built: {total_p:,} parameters.")

if NUM_GPUS > 1:
    print(f"⚡ Activating torch.nn.DataParallel across {NUM_GPUS} GPUs!")
    model = nn.DataParallel(raw_model)
else:
    model = raw_model
```

```python
# Cell 5: Strict 50-Epoch Dual GPU Training Loop with Action Conditioning
EPOCHS = 50
LR = 2e-4
optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=1e-4)
scaler = torch.amp.GradScaler('cuda', enabled=(DEVICE.type == 'cuda'))

start_epoch = 1
best_loss = float("inf")
history = []

resume_ckpt = CHECKPOINT_DIR / "last_action_diffusion_50ep.pt"
if resume_ckpt.exists():
    print(f"🔄 Found existing checkpoint at {resume_ckpt}! Resuming...")
    ckpt = torch.load(resume_ckpt, map_location=DEVICE)
    state_dict = ckpt["model_state_dict"]
    if NUM_GPUS > 1 and not list(state_dict.keys())[0].startswith("module."):
        state_dict = {"module." + k: v for k, v in state_dict.items()}
    elif NUM_GPUS <= 1 and list(state_dict.keys())[0].startswith("module."):
        state_dict = {k.replace("module.", ""): v for k, v in state_dict.items()}
    model.load_state_dict(state_dict)
    optimizer.load_state_dict(ckpt["optimizer_state_dict"])
    start_epoch = ckpt.get("epoch", 0) + 1
    best_loss = ckpt.get("loss", float("inf"))
    if (WORK / "action_diffusion_50ep_history.csv").exists():
        history = pd.read_csv(WORK / "action_diffusion_50ep_history.csv").to_dict("records")
    print(f"✅ Resumed from Epoch {start_epoch}! Best loss: {best_loss:.5f}")
else:
    print("✨ Starting brand-new Action-Conditioned Diffusion training from Epoch 01.")

print(f"🚀 Training 50 Strict Epochs on {DEVICE} (Batch Size = {BATCH_SIZE})...")
start_time = time.time()

for epoch in range(start_epoch, EPOCHS + 1):
    t0 = time.time()
    model.train()
    epoch_loss_sum, epoch_batches = 0.0, 0
    
    for batch in train_loader:
        ref = batch["reference"].to(DEVICE, non_blocking=True)
        x0 = batch["target"].to(DEVICE, non_blocking=True)
        act_id = batch["action_id"].to(DEVICE, non_blocking=True)
        b = x0.size(0)
        
        t = torch.randint(0, TIMESTEPS, (b,), device=DEVICE).long()
        noise = torch.randn_like(x0)
        
        a_bar = alphas_cumprod[t][:, None, None, None]
        xt = torch.sqrt(a_bar) * x0 + torch.sqrt(1.0 - a_bar) * noise
        
        optimizer.zero_grad(set_to_none=True)
        with torch.autocast(device_type=DEVICE.type, enabled=(DEVICE.type == 'cuda')):
            pred_noise = model(xt, t, ref, act_id)
            loss = F.mse_loss(pred_noise, noise)
            
        if not torch.isfinite(loss):
            continue
            
        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        scaler.step(optimizer)
        scaler.update()
        
        epoch_loss_sum += loss.item() * b
        epoch_batches += b
        
    train_loss = epoch_loss_sum / max(1, epoch_batches)
    history.append({"epoch": epoch, "diffusion_loss": train_loss, "lr": LR})
    ep_dur = time.time() - t0
    
    print(f"Epoch [{epoch:02d}/{EPOCHS:02d}] - Loss: {train_loss:.5f} - Time: {ep_dur:.1f}s")
    
    # Save checkpoints each epoch
    payload = {
        "epoch": epoch,
        "loss": train_loss,
        "model_state_dict": model.module.state_dict() if NUM_GPUS > 1 else model.state_dict(),
        "optimizer_state_dict": optimizer.state_dict()
    }
    torch.save(payload, CHECKPOINT_DIR / "last_action_diffusion_50ep.pt")
    pd.DataFrame(history).to_csv(WORK / "action_diffusion_50ep_history.csv", index=False)
    
    if train_loss < best_loss:
        best_loss = train_loss
        torch.save(payload, CHECKPOINT_DIR / "best_action_diffusion_50ep.pt")

total_min = (time.time() - start_time) / 60.0
print(f"\n🎉 50-Epoch Action Diffusion Completed in {total_min:.2f} minutes! Best Loss: {best_loss:.5f}")
```

```python
# Cell 6: Generate Multi-Action Verification Grid
# Synthesizes 4 DIFFERENT Actions (Walk, Slash, Thrust, Spellcast) from the SAME character!
model.eval()
batch = next(iter(train_loader))
test_char = batch["reference"][0:1].to(DEVICE) # Pick first character

demo_actions = [0, 4, 8, 12] # Spellcast_Up, Thrust_Up, Walk_Up, Slash_Up
demo_names = ["Spellcast Stance", "Thrust Attack", "Walk Forward", "Slash Strike"]

generated_frames = []

for act_id in demo_actions:
    act_tensor = torch.tensor([act_id], device=DEVICE, dtype=torch.long)
    # Controlled DDPM synthesis from reference
    t_start = 120
    n = torch.randn_like(test_char)
    a_b = alphas_cumprod[t_start]
    cur_x = torch.sqrt(a_b) * test_char + torch.sqrt(1.0 - a_b) * n
    
    with torch.no_grad():
        for step in reversed(range(t_start + 1)):
            t_t = torch.full((1,), step, device=DEVICE, dtype=torch.long)
            p_eps = model(cur_x, t_t, test_char, act_tensor)
            b_s = betas[step]
            a_s = alphas[step]
            m_s = (1.0 / torch.sqrt(a_s)) * (cur_x - (b_s / torch.sqrt(1.0 - alphas_cumprod[step])) * p_eps)
            if step > 0:
                cur_x = m_s + torch.sqrt(b_s) * torch.randn_like(cur_x) * 0.15
            else:
                cur_x = m_s
                
    frame_np = ((cur_x.clamp(-1, 1).cpu().permute(0, 2, 3, 1).numpy()[0] + 1.0) / 2.0).clip(0, 1)
    generated_frames.append(frame_np)

fig, axes = plt.subplots(1, 5, figsize=(15, 3.5), dpi=300)
ref_np = ((test_char.cpu().permute(0, 2, 3, 1).numpy()[0] + 1.0) / 2.0).clip(0, 1)
axes[0].imshow(ref_np)
axes[0].axis("off")
axes[0].set_title("Input Character", fontsize=11, fontweight="bold")

for i in range(4):
    axes[i+1].imshow(generated_frames[i])
    axes[i+1].axis("off")
    axes[i+1].set_title(demo_names[i], fontsize=11, fontweight="bold", color="#2563eb")

plt.suptitle("GameForge AI: Conditioned Multi-Action Pose Synthesis from Single Character", fontsize=13, fontweight="bold")
plt.tight_layout()
plt.savefig(WORK / "action_conditioned_verification_grid.png", dpi=300)
plt.show()
print("Saved multi-action verification grid!")
```
